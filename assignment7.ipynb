```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load sample image automatically
url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
import urllib.request
urllib.request.urlretrieve(url, "input.jpg")

img = cv2.imread("input.jpg", 0)

# Apply Laplacian filter
laplacian = cv2.Laplacian(img, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

# Sharpen image
sharp = cv2.subtract(img, laplacian)

# Display results
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sharp, cmap="gray")
plt.title("Sharpened Image")
plt.axis("off")

plt.show()
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load sample image
url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
import urllib.request
urllib.request.urlretrieve(url, "input.jpg")

img = cv2.imread("input.jpg", 0)

# Apply Sobel operators
sx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sy = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

# Calculate gradient magnitude
gradient = cv2.magnitude(sx, sy)
gradient = cv2.convertScaleAbs(gradient)

# Sharpen using Sobel gradient
sharp = cv2.add(img, gradient)

# Apply Canny edge detector
canny = cv2.Canny(img, 100, 200)

# Display results
plt.figure(figsize=(14, 4))

plt.subplot(1, 4, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(gradient, cmap="gray")
plt.title("Sobel Gradient")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(sharp, cmap="gray")
plt.title("Sobel Sharpening")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(canny, cmap="gray")
plt.title("Canny Edges")
plt.axis("off")

plt.show()
```

```python

import cv2
import matplotlib.pyplot as plt
import urllib.request

# Load sample image
url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
urllib.request.urlretrieve(url, "input.jpg")

img = cv2.imread("input.jpg", 0)

# Apply Gaussian filter (5x5, sigma=3)
blurred = cv2.GaussianBlur(img, (5, 5), 3)

# Create unsharp mask
mask = cv2.subtract(img, blurred)

# Sharpen image
sharp = cv2.add(img, mask)

# Display results
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(blurred, cmap="gray")
plt.title("Gaussian Blurred")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(sharp, cmap="gray")
plt.title("Unsharp Masking")
plt.axis("off")

plt.show()
```
