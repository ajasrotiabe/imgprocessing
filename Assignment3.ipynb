```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib.request

# Download sample image
url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
urllib.request.urlretrieve(url, "input.jpg")

# Read grayscale image
img = cv2.imread("input.jpg", 0)

# Create binary image
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

# User inputs
width = int(input("Enter border width in pixels: "))
color = int(input("Enter border color (0-255): "))

if width < 1 or not 0 <= color <= 255:
    raise ValueError("Width must be positive and color must be 0-255")

# Add border to grayscale image
gray_border = cv2.copyMakeBorder(
    img, width, width, width, width,
    cv2.BORDER_CONSTANT, value=color
)

# Add border to binary image
binary_border = cv2.copyMakeBorder(
    binary, width, width, width, width,
    cv2.BORDER_CONSTANT, value=color
)

# Display results
plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gray_border, cmap="gray", vmin=0, vmax=255)
plt.title("Grayscale with Border")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(binary, cmap="gray")
plt.title("Original Binary")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(binary_border, cmap="gray", vmin=0, vmax=255)
plt.title("Binary with Border")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib.request

# Download sample image
url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
urllib.request.urlretrieve(url, "input.jpg")

# Read grayscale image
img = cv2.imread("input.jpg", 0)

# Create binary image
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

# Calculate image complements
gray_complement = 255 - img
binary_complement = 255 - binary

# Display results
plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gray_complement, cmap="gray")
plt.title("Grayscale Complement")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(binary, cmap="gray")
plt.title("Original Binary")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(binary_complement, cmap="gray")
plt.title("Binary Complement")
plt.axis("off")

plt.tight_layout()
plt.show()
```
