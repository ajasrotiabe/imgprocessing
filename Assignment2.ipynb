```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib.request

# Download sample image
url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
urllib.request.urlretrieve(url, "input.jpg")

# Read RGB image
img = cv2.imread("input.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Separate RGB planes
R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]

# Calculate mean average
gray = ((R.astype(float) + G + B) / 3).astype(np.uint8)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original RGB Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale using Mean")
plt.axis("off")

plt.show()
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib.request

# Download sample image
url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
urllib.request.urlretrieve(url, "input.jpg")

# Read RGB image
img = cv2.imread("input.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Input weights
wr = float(input("Enter weight for Red: "))
wg = float(input("Enter weight for Green: "))
wb = float(input("Enter weight for Blue: "))

# Validate weights
if not all(0 <= w <= 1 for w in [wr, wg, wb]):
    raise ValueError("Each weight must be between 0 and 1")

if not np.isclose(wr + wg + wb, 1.0):
    raise ValueError("Sum of weights must equal 1")

# Separate RGB planes
R = img[:, :, 0].astype(float)
G = img[:, :, 1].astype(float)
B = img[:, :, 2].astype(float)

# Calculate weighted grayscale image
gray = (wr * R + wg * G + wb * B).astype(np.uint8)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original RGB Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray, cmap="gray")
plt.title("Weighted Grayscale Image")
plt.axis("off")

plt.show()
```
