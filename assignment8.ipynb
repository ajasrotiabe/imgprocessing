```python

import numpy as np
import matplotlib.pyplot as plt

N = 100000
a = 0
b = 1

# Generate using PDF-derived equation
U = np.random.random(N)
uniform_manual = a + (b - a) * U

# Generate using inbuilt function
uniform_builtin = np.random.uniform(a, b, N)

# Plot histograms
plt.figure(figsize=(10, 5))

plt.hist(uniform_manual, bins=50, alpha=0.6,
         density=True, label='PDF Equation')

plt.hist(uniform_builtin, bins=50, alpha=0.6,
         density=True, label='Inbuilt Function')

plt.title("Uniform Noise Distribution")
plt.xlabel("Noise Value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

N = 100000
mu = 0
sigma = 1

# Generate using PDF-derived Box-Muller equation
U1 = np.random.random(N)
U2 = np.random.random(N)

gaussian_manual = mu + sigma * np.sqrt(-2 * np.log(U1)) * np.cos(2 * np.pi * U2)

# Generate using inbuilt function
gaussian_builtin = np.random.normal(mu, sigma, N)

# Plot histograms
plt.figure(figsize=(10, 5))

plt.hist(gaussian_manual, bins=50, alpha=0.6,
         density=True, label='PDF Equation')

plt.hist(gaussian_builtin, bins=50, alpha=0.6,
         density=True, label='Inbuilt Function')

plt.title("Gaussian Noise Distribution")
plt.xlabel("Noise Value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

N = 100000
k = 3
lam = 2

# Generate using PDF-derived equation
U = np.random.random((N, k))

erlang_manual = -np.sum(np.log(U), axis=1) / lam

# Generate using inbuilt function
erlang_builtin = np.random.gamma(shape=k, scale=1/lam, size=N)

# Plot histograms
plt.figure(figsize=(10, 5))

plt.hist(erlang_manual, bins=50, alpha=0.6,
         density=True, label='PDF Equation')

plt.hist(erlang_builtin, bins=50, alpha=0.6,
         density=True, label='Inbuilt Function')

plt.title("Erlang Noise Distribution")
plt.xlabel("Noise Value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

N = 100000
lam = 2

# Generate using PDF-derived equation
U = np.random.random(N)

exponential_manual = -np.log(U) / lam

# Generate using inbuilt function
exponential_builtin = np.random.exponential(scale=1/lam, size=N)

# Plot histograms
plt.figure(figsize=(10, 5))

plt.hist(exponential_manual, bins=50, alpha=0.6,
         density=True, label='PDF Equation')

plt.hist(exponential_builtin, bins=50, alpha=0.6,
         density=True, label='Inbuilt Function')

plt.title("Exponential Noise Distribution")
plt.xlabel("Noise Value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

N = 100000
sigma = 1

# Generate using PDF-derived equation
U = np.random.random(N)

rayleigh_manual = sigma * np.sqrt(-2 * np.log(U))

# Generate using inbuilt function
rayleigh_builtin = np.random.rayleigh(scale=sigma, size=N)

# Plot histograms
plt.figure(figsize=(10, 5))

plt.hist(rayleigh_manual, bins=50, alpha=0.6,
         density=True, label='PDF Equation')

plt.hist(rayleigh_builtin, bins=50, alpha=0.6,
         density=True, label='Inbuilt Function')

plt.title("Rayleigh Noise Distribution")
plt.xlabel("Noise Value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid()
plt.show()
```
