```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib.request

# Download sample image
url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
urllib.request.urlretrieve(url, "input.jpg")

# Read image in grayscale
img = cv2.imread("input.jpg", 0)

# Calculate mean intensity
threshold = np.mean(img)
print("Mean intensity threshold:", threshold)

# Convert grayscale to binary
binary = np.where(img > threshold, 255, 0).astype(np.uint8)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image (Mean Threshold)")
plt.axis("off")

plt.show()
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib.request

# Download sample image
url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
urllib.request.urlretrieve(url, "input.jpg")

# Read image in grayscale
img = cv2.imread("input.jpg", 0)

# Input threshold
threshold = int(input("Enter threshold value (0-255): "))

if not 0 <= threshold <= 255:
    raise ValueError("Threshold must be between 0 and 255")

# Convert grayscale to binary
binary = np.where(img > threshold, 255, 0).astype(np.uint8)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image (User Threshold)")
plt.axis("off")

plt.show()
```
